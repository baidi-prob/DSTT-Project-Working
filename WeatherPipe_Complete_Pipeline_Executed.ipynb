### Step 2: Automated Time Alignment, Physical Range Cleaning, Causal Diurnal Imputation & Feature Engineering
- **Time Grid Alignment (`align_time_series`)**: Rounds timestamps to `1h`, averages duplicates, and reindexes to a strict hourly grid.
- **Physical Sanity & Spike Filter (`clean_physical_ranges`)**: Removes sentinel values (`-999`, `9999`, pressure `<= 500 hPa`) and single-hour hardware spikes ($>10^\circ\text{C}$ jump that immediately returns to normal).
- **Causal Diurnal Imputer (`CausalDiurnalImputer`)**: $\hat{x}_t = \mu(m_t, h_t) + 0.90^k \cdot (x_{t-k} - \mu(m_{t-k}, h_{t-k}))$
- **Domain Feature Engineering (`FeatureEngineer`)**: Generates multi-scale lags (`1, 2, 3, 6, 12, 24h`), backward rolling stats (`3h, 24h`), circular wind (`wd_sin, wd_cos`), cyclical calendar encodings, Magnus-Tetens Dew Point ($T_{\text{dew}}$), Dew Point Spread, and 3-Hour Pressure Tendency ($\Delta P_{3h}$), then prunes collinear features with $|r| > 0.98$.

```python
from weatherpipe.loader import load_raw_city_data
from weatherpipe.preprocessing import (
    align_time_series,
    clean_physical_ranges,
    CausalDiurnalImputer,
    encode_circular_wind,
)
from weatherpipe.features import FeatureEngineer

df_raw = load_raw_city_data('San Francisco', raw_dir='data/raw')
df_clean = clean_physical_ranges(align_time_series(df_raw))

imputer = CausalDiurnalImputer(decay_rate=0.90)
df_imputed = imputer.fit_transform(df_clean)
df_wind = encode_circular_wind(df_imputed)

fe = FeatureEngineer(target_col='temperature', horizon=6, corr_threshold=0.98)
X_auto, y_auto = fe.fit_transform(df_wind)

print(f"Input rows aligned:         {len(df_clean):,} hours (0 remaining NaNs after CausalDiurnalImputer)")
print(f"Total engineered features:  {X_auto.shape[1]} strictly causal features")
print(f"Collinear features pruned:  {len(fe.dropped_collinear_cols_)} (|r| > 0.98 on training split)")
print(f"Target variable:            6-hour ahead temperature T_{{t+6}}")
```
